```python
from locallib.picarrodb import *
from locallib.pandas import *
from locallib.query import *
from PostGISConnection import *
```

```python
# PostgreSQL lowercases unquoted identifiers: KPI_Data -> kpi_data (not found).
# Quote mixed-case table/column names created with "KPI_Data".
df_kpi_data = Query('SELECT * FROM kpihub."KPI_Data" WHERE customerid IN (SELECT customerid FROM kpihub."KPI_Customer" WHERE "Name" = \'ENBW\' AND "Year" = 2026)').execute(KPIHubPG_Conn)
df_kpi_data
```
