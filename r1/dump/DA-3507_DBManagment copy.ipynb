```python
import sqlite3
import pandas as pd
import numpy as np
import os
import sys
```

```python
# Create (or connect to) a SQLite3 database called "picarro_kpi.db"
db_path = os.path.join(os.getcwd(), "database", "picarro_kpi.db")
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# Optionally, check if the connection is successful
print(f"Connected to database at: {db_path}")
```

```python
# Query data from the SurveyH3 table and display the results as a DataFrame
surveyh3_df = pd.read_sql_query("SELECT * FROM SurveyH3Boundary", conn)
even_counts = surveyh3_df[surveyh3_df['Counts'] % 2 == 0]['Counts'].sum()
total_counts = surveyh3_df['Counts'].sum()
even_counts_percentage = 100 * even_counts / total_counts if total_counts != 0 else 0
even_counts_percentage
```
