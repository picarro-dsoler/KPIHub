```python
import sqlite3
import pandas as pd
import numpy as np
import os
import sys
```

```python
# Create (or connect to) a SQLite3 database called "picarro_kpi.db"
db_path = os.path.join(os.getcwd(), "picarro_kpi.db")
conn = sqlite3.connect(db_path)
cursor = conn.cursor()

# Optionally, check if the connection is successful
print(f"Connected to database at: {db_path}")
```

```python
#Create a table called KPI_Period if it doesn't already exist
cursor.execute("""
CREATE TABLE IF NOT EXISTS KPI_Period (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    value TEXT,
    description TEXT
)
```

```python
# Create a table called KPI_CONFIG if it doesn't already exist
cursor.execute("""
CREATE TABLE IF NOT EXISTS KPI_CONFIG (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    value TEXT,
    description TEXT
)
""")
conn.commit()
print("KPI_CONFIG table created or already exists.")
```

```python
# Create a table called Customers if it doesn't already exist
cursor.execute("""
CREATE TABLE IF NOT EXISTS Customers (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    region TEXT,
    CustomerId INTEGER
)
""")
conn.commit()
print("Customers table created or already exists.")
```

```python

# Create a table called Year_KPI if it doesn't already exist
cursor.execute("""
CREATE TABLE IF NOT EXISTS Y_KPI (
    ID INTEGER PRIMARY KEY AUTOINCREMENT,
    KPIId TEXT,
    KPIDefinitionId TEXT,
    CustomerId,
    Year INTEGER,
    Label TEXT,
    Value FLOAT,
    LastUpdated TIMESTAMP,
    UniqueIdentifier TEXT
)
""")
conn.commit()
print("Year_KPI table created or already exists.")

# Create a table called Weekly_KPI if it doesn't already exist (copied from Year_KPI, but with an additional Week field)
cursor.execute("""
CREATE TABLE IF NOT EXISTS W_KPI (
    ID INTEGER PRIMARY KEY AUTOINCREMENT,
    KPIId TEXT,
    KPIDefinitionId TEXT,
    CustomerId,
    Year INTEGER,
    Week INTEGER,
    Label TEXT,
    Value FLOAT,
    LastUpdated TIMESTAMP,
    UniqueIdentifier TEXT
)
""")
conn.commit()
print("Weekly_KPI table created or already exists.")

```
